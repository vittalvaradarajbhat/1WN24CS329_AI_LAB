```python
# 8 Puzzle using DFS

def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i+1], state[i+2])
    print()


def get_moves(state):
    moves = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    # Move order: Down, Right, Up, Left
    directions = [
        (1, 0, "Down"),
        (0, 1, "Right"),
        (-1, 0, "Up"),
        (0, -1, "Left")
    ]

    for dr, dc, action in directions:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            # Swap blank with tile
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            moves.append((tuple(new_state), action))

    return moves


def dfs(start, goal):

    stack = [(start, [])]
    visited = set()

    while stack:

        state, path = stack.pop()

        if state in visited:
            continue

        visited.add(state)

        # Goal test
        if state == goal:
            return path

        # Add moves to stack
        neighbors = get_moves(state)

        # Reverse because stack is LIFO
        for new_state, action in reversed(neighbors):

            if new_state not in visited:
                stack.append((new_state, path + [action]))

    return None


# -------------------------
# MAIN PROGRAM
# -------------------------

start = (
    1, 2, 3,
    4, 0, 6,
    7, 5, 8
)

goal = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

print("Initial State:")
print_puzzle(start)

print("Goal State:")
print_puzzle(goal)

# Run DFS
solution = dfs(start, goal)

if solution:

    print("Solution Found!")
    print("Number of steps:", len(solution))
    print("Moves:", " -> ".join(solution))

    # Show each step
    current = start

    print("\nSteps:")
    print("Initial State:")
    print_puzzle(current)

    for i, move in enumerate(solution):

        for new_state, action in get_moves(current):

            if action == move:
                current = new_state
                break

        print("Step", i + 1, ":", move)
        print_puzzle(current)

else:
    print("No solution found.")
```

```python
# 8 Puzzle using Iterative Deepening Search (IDS)

def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


def get_moves(state):
    moves = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    # Down, Right, Up, Left
    directions = [
        (1, 0, "Down"),
        (0, 1, "Right"),
        (-1, 0, "Up"),
        (0, -1, "Left")
    ]

    for dr, dc, action in directions:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            # Swap blank with tile
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            moves.append((tuple(new_state), action))

    return moves


# Depth Limited Search
def dls(state, goal, depth, path, visited):

    if state == goal:
        return path

    if depth == 0:
        return None

    visited.add(state)

    for new_state, action in get_moves(state):

        if new_state not in visited:

            result = dls(
                new_state,
                goal,
                depth - 1,
                path + [action],
                visited
            )

            if result is not None:
                return result

    visited.remove(state)

    return None


# Iterative Deepening Search
def ids(start, goal):

    depth = 0

    while True:

        print("Searching at depth:", depth)

        visited = set()

        result = dls(
            start,
            goal,
            depth,
            [],
            visited
        )

        if result is not None:
            return result

        depth += 1


# -------------------------
# MAIN PROGRAM
# -------------------------

start = (
    1, 2, 3,
    4, 0, 6,
    7, 5, 8
)

goal = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

print("Initial State:")
print_puzzle(start)

print("Goal State:")
print_puzzle(goal)

# Run IDS
solution = ids(start, goal)

print("Solution Found!")
print("Number of steps:", len(solution))
print("Moves:", " -> ".join(solution))

# Display solution
current = start

print("\nSolution Steps:")
print("Initial:")
print_puzzle(current)

for i, move in enumerate(solution):

    for new_state, action in get_moves(current):

        if action == move:
            current = new_state
            break

    print("Step", i + 1, ":", move)
    print_puzzle(current)

```
