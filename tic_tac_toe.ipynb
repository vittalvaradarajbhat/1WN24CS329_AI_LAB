```python
import math
import os

# Initialize the board
def create_board():
    return [[' ' for _ in range(3)] for _ in range(3)]

# Print the board to the console
def print_board(board):
    os.system('cls' if os.name == 'nt' else 'clear')
    print("\n--- Tic-Tac-Toe ---")
    for i, row in enumerate(board):
        print(f" {row[0]} | {row[1]} | {row[2]} ")
        if i < 2:
            print("-----------")
    print()

# Check if there are empty spaces left
def is_moves_left(board):
    for row in board:
        if ' ' in row:
            return True
    return False

# Evaluate the board to see if someone won
def evaluate(board):
    # Check rows for a win
    for row in board:
        if row[0] == row[1] == row[2] and row[0] != ' ':
            return 10 if row[0] == 'O' else -10

    # Check columns for a win
    for col in range(3):
        if board[0][col] == board[1][col] == board[2][col] and board[0][col] != ' ':
            return 10 if board[0][col] == 'O' else -10

    # Check diagonals for a win
    if board[0][0] == board[1][1] == board[2][2] and board[0][0] != ' ':
        return 10 if board[0][0] == 'O' else -10
    if board[0][2] == board[1][1] == board[2][0] and board[0][2] != ' ':
        return 10 if board[0][2] == 'O' else -10

    # No winner yet
    return 0

# The Minimax algorithm
def minimax(board, depth, is_max):
    score = evaluate(board)

    # Base cases: Win, Loss, or Draw
    if score == 10:
        return score - depth  # Subtract depth to favor faster wins
    if score == -10:
        return score + depth  # Add depth to favor longer survival
    if not is_moves_left(board):
        return 0

    # If it is the Maximizer's turn (AI - 'O')
    if is_max:
        best = -math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    board[i][j] = 'O'
                    best = max(best, minimax(board, depth + 1, False))
                    board[i][j] = ' '  # Undo move
        return best

    # If it is the Minimizer's turn (Player - 'X')
    else:
        best = math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    board[i][j] = 'X'
                    best = min(best, minimax(board, depth + 1, True))
                    board[i][j] = ' '  # Undo move
        return best

# Find the best possible move for the AI
def find_best_move(board):
    best_val = -math.inf
    best_move = (-1, -1)

    for i in range(3):
        for j in range(3):
            if board[i][j] == ' ':
                board[i][j] = 'O'
                move_val = minimax(board, 0, False)
                board[i][j] = ' '  # Undo move

                if move_val > best_val:
                    best_move = (i, j)
                    best_val = move_val

    return best_move

# Main game loop
def play_game():
    board = create_board()
    print_board(board)

    # Human goes first as 'X'
    print("You are 'X'. Enter row and column (1-3) separated by space (e.g., '1 1' for top-left):")

    while is_moves_left(board) and evaluate(board) == 0:
        # --- Human Turn ---
        try:
            user_input = input("Your move: ").split()
            row, col = int(user_input[0]) - 1, int(user_input[1]) - 1

            if board[row][col] != ' ':
                print("That spot is taken! Try again.")
                continue
            board[row][col] = 'X'
        except (ValueError, IndexError):
            print("Invalid input. Please enter two numbers between 1 and 3 (e.g., '2 3').")
            continue

        print_board(board)

        # Check if human won or game drew
        if evaluate(board) == -10:
            print("Congratulations! You won! (This shouldn't happen unless the AI broke!)")
            break
        if not is_moves_left(board):
            print("It's a draw!")
            break

        # --- AI Turn ---
        print("AI is thinking...")
        ai_row, ai_col = find_best_move(board)
        board[ai_row][ai_col] = 'O'

        print_board(board)

        # Check if AI won or game drew
        if evaluate(board) == 10:
            print("AI wins! Better luck next time.")
            break
        if not is_moves_left(board):
            print("It's a draw!")
            break

if __name__ == "__main__":
    play_game()
```
