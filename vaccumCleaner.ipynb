```python
# Vacuum Cleaner Agent - 2 Room Environment

# Initial state of the rooms
room = {
    'A': 'Dirty',
    'B': 'Dirty'
}

# Initial position of vacuum cleaner
position = 'A'

while True:
    print("\nCurrent Room:", position)
    print("Room A:", room['A'])
    print("Room B:", room['B'])

    if room[position] == 'Dirty':
        print("Action: SUCK")
        room[position] = 'Clean'

    elif position == 'A':
        print("Action: MOVE RIGHT")
        position = 'B'

    elif position == 'B':
        print("Action: MOVE LEFT")
        position = 'A'

    # Stop when both rooms are clean
    if room['A'] == 'Clean' and room['B'] == 'Clean':
        print("\nBoth rooms are clean!")
        print("Goal achieved.")
        break
```
